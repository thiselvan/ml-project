```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing(as_frame=True)
df = housing.frame

# Matplotlib — histogram with full control

fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(df["MedHouseVal"], bins=40, color="steelblue", edgecolor="white")
ax.set_title("Median House Value Distribution")
ax.set_xlabel("Value (x$100k)")
ax.set_ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
sns.scatterplot(data=df, x="MedInc", y="MedHouseVal", alpha=0.3)

plt.title("Income vs House Value")

plt.show()
```

```python
fig = px.scatter(df, x="MedInc", y="MedHouseVal",

                 opacity=0.3,

                 title="Income vs House Value (Interactive)",

                 labels={"MedInc": "Median Income",

                         "MedHouseVal": "House Value"})

fig.show()
```

```python
import pandas as pd

import seaborn as sns

import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame=True)

df = housing.frame

# Step 1 — Shape and types

print("Shape:", df.shape)

print(df.dtypes)

# Step 2 — Missing values

print("\nMissing values:\n", df.isnull().sum())

# Step 3 — Summary statistics

print("\nSummary:\n", df.describe().round(2))

# Step 4 — Target distribution

sns.histplot(df["MedHouseVal"], kde=True)

plt.title("Target Variable Distribution")

plt.show()

# Step 5 — Correlation heatmap

sns.heatmap(df.corr(numeric_only=True), annot=True,

            fmt=".2f", cmap="coolwarm", linewidths=0.5)

plt.title("Feature Correlations")

plt.tight_layout()

plt.show()
```
